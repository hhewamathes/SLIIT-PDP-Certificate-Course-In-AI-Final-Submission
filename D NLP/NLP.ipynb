```python
import nltk
nltk.download("punkt_tab")
from nltk.tokenize import word_tokenize

import nltk
nltk.download("stopwords")
from nltk.corpus import stopwords

from nltk.stem import PorterStemmer

import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
para = "Online shopping has become an important part of modern customer behavior. Customers regularly search for products, compare prices, read reviews, and purchase items through online platforms. Businesses collect large amounts of textual data from customer reviews, feedback, comments, and support messages. Analyzing this information can help businesses understand customer opinions, identify common problems, improve products, provide better services, and develop more effective marketing strategies that increase customer satisfaction and strengthen long-term relationships with customers."
```

```python
#Tokenization
tk_txt = word_tokenize(para)
print(tk_txt)
```

```python
#Stop Word Removal 
stop_words = set(stopwords.words("english"))

stwr_txt = []

for w in tk_txt:
    if w.lower() not in stop_words:
        stwr_txt.append(w)

print(stwr_txt)
```

```python
#Stemming
stemmer = PorterStemmer()

stmd_txt = []

for w in tk_txt:
    stmd_txt.append(stemmer.stem(w))

print(stmd_txt)
```

```python
#Lemmatization
Lemmatizer = WordNetLemmatizer()

lmtd_txt = []

for w in tk_txt:
    lmtd_txt.append(Lemmatizer.lemmatize(w))

print(lmtd_txt)
```
