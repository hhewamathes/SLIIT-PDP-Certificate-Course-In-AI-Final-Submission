```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
dtst = pd.read_csv("C:\\Users\\User\\Downloads\\SLIIP PDP Certificate Course In Artificial Intelligence\\Final Assessment\\Student_performance_data _.csv")
dtst.head()
```

```python
ss = StandardScaler()
ss_dtst = ss.fit_transform(dtst.iloc[:,[4,5,6,7,8]])
```

```python
model = KMeans(n_clusters=2)
model.fit(ss_dtst)
```

```python
model.labels_
```

```python
plt.figure(figsize=(5,5))
plt.scatter(dtst.iloc[:,4], dtst.iloc[:,13], c=model.labels_)
plt.xlabel("ParentalEducation")
plt.ylabel("GPA")
plt.title("ParentalEducation vs GPA")

plt.figure(figsize=(5,5))
plt.scatter(dtst.iloc[:,5], dtst.iloc[:,13], c=model.labels_)
plt.xlabel("StudyTimeWeekly")
plt.ylabel("GPA")
plt.title("StudyTimeWeekly vs GPA")

plt.figure(figsize=(5,5))
plt.scatter(dtst.iloc[:,6], dtst.iloc[:,13], c=model.labels_)
plt.xlabel("Absences")
plt.ylabel("GPA")
plt.title("Absences vs GPA")

plt.figure(figsize=(5,5))
plt.scatter(dtst.iloc[:,7], dtst.iloc[:,13], c=model.labels_)
plt.xlabel("Tutoring")
plt.ylabel("GPA")
plt.title("Tutoring vs GPA")

plt.figure(figsize=(5,5))
plt.scatter(dtst.iloc[:,8], dtst.iloc[:,13], c=model.labels_)
plt.xlabel("ParentalSupport")
plt.ylabel("GPA")
plt.title("ParentalSupport vs GPA")
```

```python
# ASSIGN VALUES ACCORDINGLY

ParentalEducation = float(input("ParentalEducation (0=NONE , 1=HIGH SCHOOL , 2=COLLEGE , 3=BACHELOR'S , 4=HIGHER)"))
while ParentalEducation < 0 or ParentalEducation > 4:
    print("INVALID RESPONSE")
    ParentalEducation = float(input("ParentalEducation (0=NONE , 1=HIGH SCHOOL , 2=COLLEGE , 3=BACHELOR'S , 4=HIGHER)"))

StudyTimeWeekly = float(input("StudyTimeWeekly (VALUES FROM 0-20 (HOURS))"))
while StudyTimeWeekly < 0 or StudyTimeWeekly > 20:
    print("INVALID RESPONSE")
    StudyTimeWeekly = float(input("StudyTimeWeekly (VALUES FROM 0-20 (HOURS))"))

Absences = float(input("Absences (VALUES FROM 0-30)"))
while Absences < 0 or Absences > 30:
    print("INVALID RESPONSE")
    Absences = float(input("Absences (VALUES FROM 0-30)"))

Tutoring = float(input("Tutoring (0=NO , 1=YES)"))
while Tutoring not in [0, 1]:
    print("INVALID RESPONSE")
    Tutoring = float(input("Tutoring (0=NO , 1=YES)"))

ParentalSupport = float(input("ParentalSupport (0=NONE , 1=LOW , 2=MODERATE , 3=HIGH , 4=VERY HIGH)"))
while ParentalSupport < 0 or ParentalSupport > 4:
    print("INVALID RESPONSE")
    ParentalSupport = float(input("ParentalSupport (0=NONE , 1=LOW , 2=MODERATE , 3=HIGH , 4=VERY HIGH)"))

New_Student_Data = np.array([ParentalEducation,StudyTimeWeekly,Absences,Tutoring,ParentalSupport]).reshape(1,5)
ss_New_Student_Data = ss.transform(New_Student_Data)

stdntclass = model.predict(ss_New_Student_Data)

if stdntclass == 1 :
    print("The Student Requires Minimal Teacher Guidance")
else :
    print("The Student Requires Substantial Teacher Guidance")    
```
